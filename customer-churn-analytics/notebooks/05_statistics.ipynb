# 05. Statistical Analysis

Hypothesis tests (alpha = 0.05) using SciPy/Statsmodels. Significance means an observed association in this dataset -- never causation.

```python
import pandas as pd

from src.data_cleaning import load_raw_data, clean_data
from src.segmentation import add_all_features
from src.statistics import run_all_tests, chi2_association_test, \
    charge_comparison_test, tenure_comparison_test

df = add_all_features(clean_data(load_raw_data())[0])
```

```python
chi2_association_test(df, 'Contract')
```

```python
chi2_association_test(df, 'InternetService')
```

```python
chi2_association_test(df, 'PaymentMethod')
```

```python
charge_comparison_test(df)
```

```python
tenure_comparison_test(df)
```

```python
# Full battery summary
tests = run_all_tests(df)
summary = pd.DataFrame([
    {'variable': t['variable'], 'test': t['test'],
     'statistic': round(t['statistic'], 2), 'p_value': f"{t['p_value']:.2e}",
     'df': t['degrees_of_freedom']} for t in tests
])
summary
```

## Interpretation

All five tests reject the null hypothesis of no association at alpha = 0.05:
- Contract, internet service and payment method are each associated with churn (chi-square p < 1e-140).
- Monthly charges distributions differ between churned and retained customers (Mann-Whitney U, p < 1e-53); churned customers carry higher charges on average.
- Tenure distributions differ strongly (Mann-Whitney U, p < 1e-208); churned customers have much shorter tenure on average.

These results quantify association strength only. They do not prove that any variable causes churn.
